```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/KB_FinalProject_Team6/데이터파일/학습에 쓸 데이터/카드데이터_전처리_v02.csv")
df.head()
```

```python
# 3. 행정동명을 첫 번째 공백에서 분리
region = df["hname"].str.strip().str.split(
    pat=r"\s+",
    n=1,
    expand=True,
).reindex(columns=[0, 1])
```

```python
df["구군"] = region[0]
df["읍면동"] = region[1]
```

```python
df = df.rename(columns={
    "stdYm": "기준월",
    "hcode": "행정동코드",
    "ryLNm": "카드업종",
    "avgAmt": "평균금액",
    "avgCnt": "평균건수",
})
```

```python
df = df.drop(columns="hname")
```

```python
df = df[
    [
        "기준월",
        "행정동코드",
        "구군",
        "읍면동",
        "카드업종",
        "평균금액",
        "평균건수",
    ]
]
```

```python
# 6. 결과 확인
print(f"변환 완료: {len(df):,}행")
display(df.head())
```

```python
# 7. 한글 호환 CSV로 저장하고 다운로드
output_path = "/content/카드데이터_한글컬럼.csv"
df.to_csv("/content/drive/MyDrive/KB_FinalProject_Team6/데이터파일/학습에 쓸 데이터/카드데이터_전처리_v03.csv", index=False, encoding="utf-8-sig")

```
