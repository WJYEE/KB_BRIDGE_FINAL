```python
from google.colab import drive
import pandas as pd

# Google Drive 마운트
drive.mount('/content/drive')
```

```python
df = pd.read_csv("/content/drive/MyDrive/KB_FinalProject_Team6/데이터파일/학습에 쓸 데이터/시간당 유동인구(통신데이터)_PREPROCESSED.csv")
df.head()
```

```python
required = [
    "stdYm", "hcode", "hname", "gu", "dong",
    "area_km2", "times", "hour",
    "hpopAvg", "wpopAvg", "vpopAvg", "impute_method",
]
```

```python
missing = set(required) - set(df.columns)

if missing:
    raise ValueError(f"필요한 컬럼이 없습니다: {sorted(missing)}")
```

```python
# 3. 공통 식별 컬럼을 문자열로 변환하고 앞뒤 공백 제거
for col in ["stdYm", "hcode", "gu", "dong"]:
    df[col] = df[col].astype("string").str.strip()
```

```python
# 4. 컬럼명을 모두 한글로 변경
df = df.rename(columns={
    "stdYm": "기준월",
    "hcode": "행정동코드",
    "hname": "행정동명",
    "gu": "구군",
    "dong": "읍면동",
    "area_km2": "면적_제곱킬로미터",
    "times": "시간표시",
    "hour": "시간",
    "hpopAvg": "거주인구_평균",
    "wpopAvg": "직장인구_평균",
    "vpopAvg": "방문인구_평균",
    "impute_method": "보간방법",
})
```

```python
# 5. 중복 지역명 제거 및 컬럼 순서 정리
df = df.drop(columns="행정동명")

df = df[
    [
        "기준월",
        "행정동코드",
        "구군",
        "읍면동",
        "면적_제곱킬로미터",
        "시간",
        "시간표시",
        "거주인구_평균",
        "직장인구_평균",
        "방문인구_평균",
        "보간방법",
    ]
]
```

```python
# 6. 변환 결과 확인
print(f"변환 완료: {len(df):,}행")
print("카드 데이터 공통 컬럼: 기준월, 행정동코드, 구군, 읍면동")
display(df.head())
```
